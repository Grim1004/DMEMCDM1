# Filtrado, ordenamiento, mezclas y agregación

## Filtrado de filas de acuerdo a condiciones lógicas

En ocasiones nos interesan ciertas observaciones por lo que necesitamos sera capaces de extraerlas. Veamos las distintas etapas que existen para luego filtrar las observaciones correspondientes con la etapa inicial de la enfermedad.

```python
# Número de individuos en cada etapa histológica de la cirrosis
cirrosis.Stage.value_counts() # cuenta los valores únicos en la columna 'Stage'
```

Existen 21 observaciones de personas que estan en etapa 1

```python
# Filtrar las filas donde la etapa de cirrosis es menor a 2
cirrosis_etapa_1 = cirrosis[cirrosis['Stage'] < 2] # Filtrar filas donde 'Stage' es menor a 2
cirrosis_etapa_1.head(5)
```

```python
cirrosis_etapa_1.Stage.value_counts() # cuenta los valores únicos en la columna 'Stage'
```

OJO: Cuidado con el tipo de dato que tiene cada variable porque en ocasiones al no tomarlo en consideración las operaciones y los filtros pueden no comportarse de manera adecuada.

```python
cirrosis[cirrosis['Stage'] == '1.0'] # Filtrar filas donde 'Stage' es igual a 1.0
```

En ocasiones se pueden necesitar filtros que dependan de más de una variables; por ejemplo, menores de edad en etapa 1 de género F. Hay que tener cuidado pues la edad esta en días.

```python
# Crear una máscara booleana para filtrar etapas menores a 2
stage_mask = (cirrosis['Stage'] < 2) # Filtrar etapas menores a 2
stage_mask
```

Para verificar que funciona bien la máscara, pueden sumar los boleanos y deberían ser 21

```python
# Aplicar la máscara booleana para contar cuántas etapas menores a 2 hay
stage_mask.sum()
```

## Máscaras booleanas adicionales

Es una forma mucho más sofisticada de hacer filtros

```python
# Máscaras booleanas adicionales
age_mask = (cirrosis['Age'] / 365 < 18) # Filtrar edades menores a 18 años
sex_mask = (cirrosis['Sex'] == 'F') # Filtrar solo mujeres
```

Al hacer esta división y no conocer la fecha de nacimiento introduciríamos un cierto tipo de sesgo

Filtro las observaciones que cumplan con todos los requisitos.

```python
# Encontrar mujeres menores de edad en etapas de cirrosis menores a la etapa 2
cirrosis_F_menores_edad = cirrosis[stage_mask & age_mask & sex_mask]
```

Equivalentemente podrían escribirlo como sigue

```python
# Encontrar mujeres menores de edad en etapas de cirrosis menores a la etapa 2 (sin usar máscaras intermedias)
cirrosis[(cirrosis['Stage'] < 2) & (cirrosis['Age'] / 365 < 18) & (cirrosis['Sex'] == 'F')]
```

```python
# ¿Qué resultados arrojan ambas formas de filtrar?
cirrosis_F_menores_edad
```

Observamos que en el estudio no hay mujeres menores de edad

```python
# Contar cuántas mujeres menores de edad hay en el dataset
(cirrosis['Age'] / 365 < 18).sum()
```

Revisando la documentación nos enteramos que este era un estudio clínico que se hizo para comprobar la efectividad de un tratamiento novedoso para la cirrosis por lo que no sería legal hacer estudios clínicos experimentales en niños (sin consentimiento de sus tutores legales y en casos no extremos) por lo cual nuestra base no tiene información de menores de edad.

Bueno... vamos a relajar un poco el supuesto y digamos que queremos menores de edad o mayores de 65 años.

```python
# Encontrar mujeres que no están en edad laboral (menores de 18 o mayores de 65) en etapas de cirrosis menores a la etapa 2
age_mask_2 = age_mask | (cirrosis['Age'] / 365 > 65) # máscara para menores de 18 o mayores de 65
cirrosis_F_no_laboral = cirrosis[stage_mask & sex_mask & age_mask_2]
```

Equivalentemente...

```python
# Encontrar mujeres que no están en edad laboral (menores de 18 o mayores de 65) en etapas de cirrosis menores a la etapa 2 (sin usar máscaras intermedias)
cirrosis[(cirrosis['Stage'] < 2) & ((cirrosis['Age'] / 365 < 18) | (cirrosis['Age'] / 365 > 65)) & (cirrosis['Sex'] == 'F')]
```

```python
# ¿Qué resultados arrojan ambas formas de filtrar
cirrosis_F_no_laboral
```


Observamos que en el estudio no hay mujeres menores de edad ni mujeres mayores de 65 años

Haciendo un análisis de la información vemos que el rango de edad en pacientes género 'F' que se encuentran en etapa 1 esta entre 28 y 62 años. Por lo cual hace sentido que no recuperemos información, ya que no mujeres menores de edad ni mayores a 65 años.

```python
# Revisar estadísticas descriptivas de la edad de las mujeres
(cirrosis[sex_mask].Age /365).describe()
```

```python
# Revisar estadísticas descriptivas de la edad de las mujeres en etapas menores a la etapa 2
(cirrosis[stage_mask].Age /365).describe()
```

```python
# Revisar estadísticas descriptivas de la edad de las mujeres menores de 18 o mayores de 65 en etapa 1 
(cirrosis[age_mask_2].Age /365).describe()
```

En conclusión: fracasamos rotundamente con exito. No pudimos filtrar porque no hay información pero fuimos capaces de berificar en las bases de datos que es correcto que no haya información por las caracteristicas de las bases
